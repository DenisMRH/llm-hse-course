```python
from your_solution import prepare_tokenizer

tokenizer = prepare_tokenizer()
text = "Москва — столица России."

ids = tokenizer.encode(text, add_special_tokens=False)

print("Исходный текст:", text)
print("Токены:", tokenizer.convert_ids_to_tokens(ids))
print("ID токенов:", ids)
print("Обратное преобразование:", tokenizer.decode(ids))
print("Размер словаря:", tokenizer.vocab_size)
print("EOS:", tokenizer.eos_token, tokenizer.eos_token_id)
print("PAD:", tokenizer.pad_token, tokenizer.pad_token_id)
```
